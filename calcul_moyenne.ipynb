# Calcul des moyennes d'UE

## Verion 1

```python
def v1():
    som = 0
    acc = 0
    l_mat = [f"R{i}" for i in range(101,105)] + [f"SAE{i}" for i in range(11,13)]
    for y in l_mat:
        coeff = int(input(f"entrez le coefficient de la {y}"))
        note = float(input(f"entrez la note de la {y}"))
        som +=coeff
        acc += note * coeff
    acc /= som
    print(acc)

v1()
```

```python
def v2():
    acc = 0
    l_coeff = []
    l_note = []
    l_mat = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]
    for y in l_mat:
        coeff = int(input(f"entrez le coefficient de la {y}"))
        l_coeff.append(coeff)
        note = int(input(f"entrez la note de la {y}"))
        l_note.append(note)
        acc += note * coeff
    acc /= len(l_note)
    print(acc)

v1()
```
